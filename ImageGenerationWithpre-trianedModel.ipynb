```python
import torch
from diffusers import StableDiffusionPipeline

pipe = StableDiffusionPipeline.from_pretrained(
    "stable-diffusion-v1-5/stable-diffusion-v1-5",
    torch_dtype=torch.float16
).to("cuda")
```

```python
image = pipe("a futuristic city at sunset, highly detailed, digital art",
             negative_prompt="blurry, low quality, distorted",
             num_inference_steps=30, guidance_scale=7.5).images[0]
image.save("output.png")
image
```

```python
image = pipe("a cute fluffy cat sitting on a sofa, realistic photo, highly detailed, soft lighting",
             negative_prompt="blurry, low quality, distorted",
             num_inference_steps=30, guidance_scale=7.5).images[0]
image.save("cat.png")
image
```
